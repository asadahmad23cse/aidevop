# Medical QA Fine-Tuning Analysis
This notebook analyzes dataset properties, training behavior, and evaluation outcomes for the QLoRA fine-tuning project.

## 1) Dataset Statistics
Compute dataset sizes, average lengths, and a topic distribution chart based on keyword matching.

```python
from pathlib import Path
import json
from collections import Counter
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
base = Path('..')
data_dir = base / 'data' / 'processed'

def load_jsonl(path):
    rows = []
    with path.open('r', encoding='utf-8') as f:
        for line in f:
            if line.strip():
                rows.append(json.loads(line))
    return rows

train = load_jsonl(data_dir / 'train.jsonl')
val = load_jsonl(data_dir / 'val.jsonl')
test = load_jsonl(data_dir / 'test.jsonl')
all_rows = train + val + test

avg_q_len = sum(len(r['input'].split()) for r in all_rows) / max(1, len(all_rows))
avg_a_len = sum(len(r['output'].split()) for r in all_rows) / max(1, len(all_rows))
print(f'Train: {len(train)} | Val: {len(val)} | Test: {len(test)} | Total: {len(all_rows)}')
print(f'Average question length (words): {avg_q_len:.2f}')
print(f'Average answer length (words): {avg_a_len:.2f}')

topics = ['diabetes', 'hypertension', 'asthma', 'depression', 'arthritis', 'fever', 'migraine', 'thyroid', 'anemia', 'anxiety']
counter = Counter()
for row in all_rows:
    text = (row['input'] + ' ' + row['output']).lower()
    matched = [t for t in topics if t in text]
    if matched:
        for t in matched:
            counter[t] += 1
    else:
        counter['other'] += 1

labels = list(counter.keys())
values = [counter[k] for k in labels]
plt.figure(figsize=(10, 4))
sns.barplot(x=labels, y=values, palette='crest')
plt.xticks(rotation=45, ha='right')
plt.title('Topic Distribution (Keyword-Based)')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

## 2) Training Loss Curve
Plot train loss and validation loss over epochs from `outputs/results/training_history.json`.

```python
history_path = base / 'outputs' / 'results' / 'training_history.json'
history = json.loads(history_path.read_text(encoding='utf-8'))

train_epochs = [x['epoch'] for x in history.get('train_loss', [])]
train_vals = [x['loss'] for x in history.get('train_loss', [])]
val_epochs = [x['epoch'] for x in history.get('val_loss', [])]
val_vals = [x['loss'] for x in history.get('val_loss', [])]

plt.figure(figsize=(8, 4))
if train_epochs:
    plt.plot(train_epochs, train_vals, marker='o', label='Train Loss')
if val_epochs:
    plt.plot(val_epochs, val_vals, marker='s', label='Val Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Train vs Validation Loss')
plt.legend()
plt.tight_layout()
plt.show()
```

## 3) Metrics Comparison
Compare ROUGE, BLEU, and BERTScore between base and fine-tuned models.

```python
report_path = base / 'outputs' / 'results' / 'evaluation_report.json'
report = json.loads(report_path.read_text(encoding='utf-8'))

base_metrics = report['base_model']
ft_metrics = report['fine_tuned_model']
metric_keys = ['rouge_1', 'rouge_2', 'rouge_l', 'bleu', 'bertscore_f1']

x = list(range(len(metric_keys)))
width = 0.38

plt.figure(figsize=(10, 4))
plt.bar([i - width/2 for i in x], [base_metrics[m] for m in metric_keys], width=width, label='Base')
plt.bar([i + width/2 for i in x], [ft_metrics[m] for m in metric_keys], width=width, label='Fine-Tuned')
plt.xticks(x, metric_keys, rotation=25, ha='right')
plt.title('Base vs Fine-Tuned Metrics')
plt.legend()
plt.tight_layout()
plt.show()
```

## 4) Five Qualitative Comparisons
Render five example prompt/response comparisons from `sample_outputs.txt` in a markdown table.

```python
from IPython.display import Markdown, display

sample_path = base / 'outputs' / 'results' / 'sample_outputs.txt'
text = sample_path.read_text(encoding='utf-8')
chunks = [c.strip() for c in text.split('=' * 80) if c.strip()]
rows = []
for chunk in chunks[:5]:
    lines = [ln for ln in chunk.splitlines() if ln.strip()]
    prompt = ''
    base_resp = ''
    ft_resp = ''
    mode = None
    for ln in lines:
        if ln.startswith('Prompt:'):
            mode = 'prompt'
            continue
        if ln.startswith('Base Model Response:'):
            mode = 'base'
            continue
        if ln.startswith('Fine-Tuned Model Response:'):
            mode = 'ft'
            continue
        if ln.startswith('Example '):
            continue
        if mode == 'prompt':
            prompt += (' ' + ln)
        elif mode == 'base':
            base_resp += (' ' + ln)
        elif mode == 'ft':
            ft_resp += (' ' + ln)
    rows.append((prompt.strip(), base_resp.strip(), ft_resp.strip()))

md = ['| Prompt | Base | Fine-Tuned |', '|---|---|---|']
for p, b, f in rows:
    p = p.replace('|', ' ')
    b = b.replace('|', ' ')
    f = f.replace('|', ' ')
    md.append(f'| {p[:120]} | {b[:120]} | {f[:120]} |')
display(Markdown('\n'.join(md)))
```

## 5) Written Analysis
### Methodology
We combine curated medical QA data from Hugging Face with synthetic topic-controlled examples, format data in an instruction-following template, and fine-tune with QLoRA (4-bit NF4 quantization + LoRA adapters on attention projections).

### Findings
Use the charts above to compare model quality and efficiency. Typical expected behavior is improved domain relevance and lexical overlap for the fine-tuned model, with some increase in latency from adapter overhead.

### Limitations
Evaluation metrics (ROUGE/BLEU/BERTScore) are proxy signals and do not fully capture medical safety. The dataset may contain biases or outdated medical practices, and this assistant is not a replacement for licensed medical care.
